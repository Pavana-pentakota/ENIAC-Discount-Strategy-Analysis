```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
```

1. How many products are being discounted?

```python
# A product (sku) counts as "discounted" if it was ever sold below its listed price
discount_flag = category_order.groupby('sku', as_index=False)['discount'].max()
discount_flag['discounted'] = discount_flag['discount'] > 0
discount_summary = discount_flag['discounted'].value_counts().rename({True: 'Discounted', False: 'Not discounted'})

plt.figure(figsize=(6, 5))
sns.barplot(x=discount_summary.index, y=discount_summary.values, palette=['#4C72B0', '#DD8452'])
plt.xlabel('')
plt.ylabel('Number of unique products (SKUs)')
for i, v in enumerate(discount_summary.values):
    plt.text(i, v + discount_summary.values.max() * 0.01, f'{v:,}', ha='center', fontweight='bold')
plt.tight_layout()
plt.show()

pct_discounted = 100 * discount_summary.get('Discounted', 0) / discount_summary.sum()
print(f"{discount_summary.get('Discounted', 0):,} products ({pct_discounted:.1f}%) were sold at a discount at least once.")
```

2. How big are the offered discounts as a percentage of the products' prices?

```python
discounted_lines = category_order.loc[category_order['discount'] > 0].copy()
discounted_lines['discount_pct'] = (discounted_lines['discount'] / discounted_lines['price']) * 100

# guard against corrupted/stale price data producing impossible outliers (e.g. >100% off)
discounted_lines = discounted_lines.loc[discounted_lines['discount_pct'].between(0, 100)]

plt.figure(figsize=(9, 5))
sns.histplot(discounted_lines['discount_pct'], bins=40, kde=True, color='#4C72B0')
plt.axvline(discounted_lines['discount_pct'].median(), color='red', linestyle='--',
            label=f"Median: {discounted_lines['discount_pct'].median():.1f}%")
plt.title('Distribution of Discount Size (as % of Product Price)')
plt.xlabel('Discount as % of price')
plt.ylabel('Number of order lines')
plt.legend()
plt.tight_layout()
plt.show()

discounted_lines['discount_pct'].describe()
```

3. How do seasonality and special dates (Christmas, Black Friday) affect sales?

```python
monthly_plot = monthly.copy()
monthly_plot.index = monthly_plot.index.to_timestamp()

# Black Friday (late Nov) and Christmas (Dec) peaks, for any year covered by the data
special_months = {}
for year in monthly_plot.index.year.unique():
    special_months[pd.Timestamp(f'{year}-11-01')] = 'Black Friday'
    special_months[pd.Timestamp(f'{year}-12-01')] = 'Christmas'

fig, ax = plt.subplots(figsize=(12, 6))
ax.plot(monthly_plot.index, monthly_plot['revenue'], marker='o', color='#4C72B0', label='Monthly revenue')

for ts, label in special_months.items():
    if ts in monthly_plot.index:
        ax.scatter(ts, monthly_plot.loc[ts, 'revenue'], color='red', s=100, zorder=5)
        ax.annotate(label, (ts, monthly_plot.loc[ts, 'revenue']), textcoords='offset points',
                    xytext=(0, 12), ha='center', color='red', fontweight='bold')

ax.set_title('Monthly Revenue, with Seasonal / Promotional Peaks Highlighted')
ax.set_xlabel('Month')
ax.set_ylabel('Revenue (EUR)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

4. How should products be classified into different categories to simplify reports and analysis?

```python
# The rule-based tagger can assign multiple labels to one product (e.g. 'charger_cable, keyboard'),
# which is too granular for reporting. Simplify by keeping only the first (primary) matched category.
category_order['primary_category'] = category_order['category'].str.split(',').str[0].str.strip()

category_counts = (category_order.groupby('primary_category')['sku']
                    .nunique()
                    .sort_values(ascending=False)
                    .head(15))

plt.figure(figsize=(10, 6))
sns.barplot(x=category_counts.values, y=category_counts.index, palette='viridis')
plt.title('Top 15 Product Categories by Number of Unique Products')
plt.xlabel('Number of unique products (SKUs)')
plt.ylabel('Category')
plt.tight_layout()
plt.show()
```

5. What is the distribution of product prices across different categories?

```python
top_categories = category_counts.index.tolist()
price_by_category = category_order.loc[category_order['primary_category'].isin(top_categories)].drop_duplicates(subset='sku')

order = price_by_category.groupby('primary_category')['price'].median().sort_values(ascending=False).index

plt.figure(figsize=(12, 6))
sns.boxplot(data=price_by_category, x='primary_category', y='price', order=order, palette='coolwarm')
plt.yscale('log')
plt.title('Product Price Distribution by Category (log scale)')
plt.xlabel('Category')
plt.ylabel('Price, EUR (log scale)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
